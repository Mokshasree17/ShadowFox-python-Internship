```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 18000, 15000, 22000, 27000, 25000]
}

df = pd.DataFrame(sales)
df
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(6,4))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

How did monthly sales change from january to june?

Sales generally increased from January to May, reaching the highest value of 27,000 in May. There was a slight decrease to 25,000 in June, but the overall trend remained positive.
